### Which one should you use? (pyplot vs matplotlib OO method?)
* When plotting something quickly, okay to use the pyplot method
* When plotting something more advanced, use to OO method

```python
over50 = heartDisease[heartDisease["age"]>50]
over50.head()
```

```python
len(over50)
```

```python
#Pyplot method
over50.plot(kind = "scatter",
            x = "age",
            y = "chol",
            c = "target");
```

```python
# OO method
fig, ax = plt.subplots(figsize = (10,6))
over50.plot(kind = "scatter",
            x = "age",
            y = "chol",
            c = "target",
           ax = ax);

# ax.set_xlim([45,100])
```

```python
## OO method from scratch
fig, ax = plt.subplots(figsize = (10,6))

# Plot the data
scatter = ax.scatter(x = over50["age"],
                    y = over50["chol"],
                    c = over50["target"].map({"yes":1,"no":0}));

#Customize the plot
ax.set(title = "Heart Disease and Cholestrol Levels",
       xlabel = "age",
       ylabel = "Cholestrol");

# Add a legend
ax.legend(*scatter.legend_elements(), title = "Target");

#Add a horizontal line
ax.axhline(over50["chol"].mean(),
          linestyle = '--');
```

```python
over50.head()
```

```python
# Sublpot of chol, age, thalach
fig, (ax0,ax1) = plt.subplots(nrows = 2,
                              ncols = 1,
                              figsize = (10,10),
                             sharex = True)

# Add data to a ax0
scatter0 = ax0.scatter(x = over50["age"],
            y = over50["chol"],
            c = over50["target"].map({"yes":1,"no":0}));

# Add data to ax1
scatter1 = ax1.scatter(x = over50["age"],
            y = over50["thalach"],
            c = over50["target"].map({"yes":1,"no":0}));

#Customize ax0
ax0.set(title = "Heart Disease and Cholestrol Levels",
        ylabel = "Cholestrol")

#Customize ax1
ax1.set(title = "Heart Disease and Maximum Heart Rate",
        xlabel = "Age",
        ylabel = "Maximum Heart Rate")

#Add legend to ax0
ax0.legend(*scatter0.legend_elements(),title = "Target")

#Add legend to ax1
ax1.legend(*scatter1.legend_elements(),title = "Target")

#Add a horizontal line to ax0
ax0.axhline(over50["chol"].mean(),
            linestyle = ':')

#Add a horizontal line to ax1
ax1.axhline(over50["thalach"].mean(),
            linestyle = '-.');

#Add a title to the figure
fig.suptitle("Heart Disease Analysis", fontsize = 16, fontweight = "bold");
```

```python
#Set the style
plt.style.use('seaborn-v0_8-whitegrid')
## OO method from scratch
fig, ax = plt.subplots(figsize = (10,6))

# Plot the data
scatter = ax.scatter(x = over50["age"],
                    y = over50["chol"],
                    c = over50["target"].map({"yes":1,"no":0}),
                    cmap="winter");#Changes the color scheme

#Customize the plot
ax.set(title = "Heart Disease and Cholestrol Levels",
       xlabel = "age",
       ylabel = "Cholestrol");

# Add a legend
ax.legend(*scatter.legend_elements(), title = "Target");

#Add a horizontal line
ax.axhline(over50["chol"].mean(),
          linestyle = '--');
```

```python
#Customising the x and y axis limitations
# Sublpot of chol, age, thalach
fig, (ax0,ax1) = plt.subplots(nrows = 2,
                              ncols = 1,
                              figsize = (10,10),
                             sharex = True)

# Add data to a ax0
scatter0 = ax0.scatter(x = over50["age"],
            y = over50["chol"],
            c = over50["target"].map({"yes":1,"no":0}),
                      cmap = "winter");

# Add data to ax1
scatter1 = ax1.scatter(x = over50["age"],
            y = over50["thalach"],
            c = over50["target"].map({"yes":1,"no":0}),
                      cmap = "winter");

#Customize ax0
ax0.set(title = "Heart Disease and Cholestrol Levels",
        ylabel = "Cholestrol")

#Change the x axis limits
ax0.set_xlim([50,80])

#Customize ax1
ax1.set(title = "Heart Disease and Maximum Heart Rate",
        xlabel = "Age",
        ylabel = "Maximum Heart Rate")

#Change the x and y axis limits
ax1.set_xlim([50,80])
ax1.set_ylim([60,200])

#Add legend to ax0
ax0.legend(*scatter0.legend_elements(),title = "Target")

#Add legend to ax1
ax1.legend(*scatter1.legend_elements(),title = "Target")

#Add a horizontal line to ax0
ax0.axhline(over50["chol"].mean(),
            linestyle = ':')

#Add a horizontal line to ax1
ax1.axhline(over50["thalach"].mean(),
            linestyle = '-.');

#Add a title to the figure
fig.suptitle("Heart Disease Analysis", fontsize = 16, fontweight = "bold");
```
